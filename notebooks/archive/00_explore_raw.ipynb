# 00 — Explore raw

Inspection libre d'un fichier .mat brut. Aucune écriture.
Utilité : déboguer un fichier suspect, vérifier la visibilité des marqueurs, etc.

```python
from resilience import paths, participants
from resilience.io import loader
from resilience.viz import plots
import matplotlib.pyplot as plt
%matplotlib inline

paths.check_dirs(verbose=True)
```

```python
# Choix de l'essai à inspecter
PARTICIPANT = '001CrMa'
CONDITION   = 'silence'

raw_path = paths.raw_file(PARTICIPANT, CONDITION)
print(raw_path, '→ existe :', raw_path.exists())
```

```python
data, fmt = loader.load_mat(raw_path)
trajectories = loader.extract_marker_trajectories(data, fmt)
loader.print_visibility_report(raw_path, trajectories)
```

```python
fig = plots.plot_dos_trajectories(trajectories, PARTICIPANT, CONDITION)
plt.show()
```
